```python
import pandas as pd
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
```

```python
import os
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import roc_auc_score

# ============================================================
# EXP_026 — EXTRA TREES DIVERSITY ENGINE
# ============================================================

BASE = r"C:\Users\Garv\Stats & Data Engineering\KAGGLE\ev_kaggle_artifacts"

EXP_DIR = os.path.join(
    BASE,
    "experiments",
    "EXP_026_EXTRATREES"
)

os.makedirs(EXP_DIR, exist_ok=True)

print("Train shape:", train.shape)
print("Test shape :", test.shape)

# ============================================================
# TARGET
# ============================================================

TARGET = "Will_Buy_EV"

assert TARGET in train.columns, (
    f"Target '{TARGET}' not found.\n"
    f"Columns: {train.columns.tolist()}"
)

# y = train[TARGET].astype(int)
y = train[TARGET].map({"No": 0, "Yes": 1}).astype(int)

print("Target distribution:")
print(y.value_counts())
print("\nTarget rate:", y.mean())

X = train.drop(columns=[TARGET]).copy()
X_test = test.copy()

# Remove ID from learning if present
for col in ["id", "ID"]:
    if col in X.columns:
        X = X.drop(columns=[col])
    if col in X_test.columns:
        X_test = X_test.drop(columns=[col])

# ============================================================
# COLUMN TYPES
# ============================================================

cat_cols = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

num_cols = [
    c for c in X.columns
    if c not in cat_cols
]

print("\nNumerical columns:", len(num_cols))
print("Categorical columns:", len(cat_cols))

print("\nCategorical:")
print(cat_cols)

# ============================================================
# PREPROCESSING
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            "passthrough",
            num_cols
        ),
        (
            "cat",
            OrdinalEncoder(
                handle_unknown="use_encoded_value",
                unknown_value=-1
            ),
            cat_cols
        )
    ],
    remainder="drop"
)

# ============================================================
# MODEL
# ============================================================

model = ExtraTreesClassifier(
    n_estimators=500,
    max_features="sqrt",
    min_samples_leaf=2,
    class_weight="balanced",
    n_jobs=-1,
    random_state=26026
)

pipe = Pipeline([
    ("preprocess", preprocessor),
    ("model", model)
])

# ============================================================
# 5-FOLD OOF
# ============================================================

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=26026
)

oof = np.zeros(len(X), dtype=np.float64)
test_pred = np.zeros(len(X_test), dtype=np.float64)

for fold, (tr_idx, va_idx) in enumerate(
    skf.split(X, y),
    start=1
):

    print(f"\n{'='*60}")
    print(f"FOLD {fold}/5")
    print("="*60)

    X_tr = X.iloc[tr_idx]
    X_va = X.iloc[va_idx]

    y_tr = y.iloc[tr_idx]

    pipe.fit(X_tr, y_tr)

    oof[va_idx] = pipe.predict_proba(X_va)[:, 1]

    test_pred += (
        pipe.predict_proba(X_test)[:, 1] / skf.n_splits
    )

    fold_auc = roc_auc_score(
        y.iloc[va_idx],
        oof[va_idx]
    )

    print("Fold AUC:", fold_auc)

# ============================================================
# FINAL OOF SCORE
# ============================================================

auc = roc_auc_score(y, oof)

print("\n")
print("=" * 70)
print("EXP_026 RESULTS")
print("=" * 70)

print("OOF AUC:", auc)

# ============================================================
# SAVE
# ============================================================

np.save(
    os.path.join(EXP_DIR, "oof.npy"),
    oof
)

np.save(
    os.path.join(EXP_DIR, "test.npy"),
    test_pred
)

pd.DataFrame({
    "metric": ["OOF_AUC"],
    "value": [auc]
}).to_csv(
    os.path.join(EXP_DIR, "metrics.csv"),
    index=False
)

print("\nSaved:")
print(os.path.join(EXP_DIR, "oof.npy"))
print(os.path.join(EXP_DIR, "test.npy"))
print(os.path.join(EXP_DIR, "metrics.csv"))
```
